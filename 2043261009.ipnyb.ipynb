```python
bulan=input("Bulan menginap (Mei, Juni, Juli, Agustus, September, Oktober): ")
lama=int(input("Lama menginap (hari): "))

studio=0
apartemen=0
if bulan == 'Mei' or bulan == 'Oktober':
  studio=500000
  apartemen=650000
elif bulan == 'Juni' or bulan == 'September':
  studio=752000
  apartemen=687000
elif bulan == 'Juli' or bulan == 'Agustus':
  studio=760000
  apartemen=770000
else:
  print('Bulan tidak tersedia.')

diskon=0
if lama > 3:
  diskon = 0.15
else:
  diskon = 0

total_studio=studio*lama*(1-diskon)
total_apartemen=apartemen*lama*(1-diskon)

print("Studio: Rp",total_studio)
print("Apartemen: Rp",total_apartemen)
```

```python
nama_barang=input('Nama barang: ') #a
jumlah_barang=int(input('Jumlah barang: '))
satuan_awal=input('Satuan jumlah tersebut (pcs/lusin/kodi): ')
satuan_tujuan=input('Satuan tujuan (pcs/lusin/kodi): ')


if satuan_awal == 'kodi':
  total_pcs=jumlah_barang*20
elif satuan_awal == 'lusin':
  total_pcs=jumlah_barang*12
else:
  print('Satuan tidak valid.')


satuan_akhir=0
if satuan_awal == 'pcs' and satuan_tujuan == 'lusin':
    satuan_akhir= jumlah_barang/12
elif satuan_awal == 'lusin' and satuan_tujuan == 'pcs':
    satuan_akhir= jumlah_barang*12
elif satuan_awal == 'kodi' and satuan_tujuan == 'pcs':
    satuan_akhir= jumlah_barang*20
elif satuan_awal == 'pcs' and satuan_tujuan == 'kodi':
    satuan_akhir= jumlah_barang/20
elif satuan_awal == 'lusin' and satuan_tujuan == 'kodi':
    satuan_akhir= jumlah_barang*0.6
elif satuan_awal == 'kodi'and satuan_tujuan == 'lusin':
    satuan_akhir= jumlah_barang*1.6
else:
  print('Satuan tidak tersedia.')

print('')
print('Barang:', nama_barang)
print('Jumlah awal:', jumlah_barang, satuan_awal)
print('Total dalam pcs:', total_pcs, 'pcs')
print('Konversi ke:', round(satuan_akhir,2), satuan_tujuan)
```
